```python
import json
import jsonschema

# 載入官方 schema 與你的 D-Plan
with open("D-Plan.schema.json", "r", encoding="utf-8") as f:
    schema = json.load(f)

with open("D-Plan_TEAM_11402_2026-10-27.json", "r", encoding="utf-8") as f:
    dplan = json.load(f)

# 執行第一層結構驗證（出錯會噴出具體哪個欄位不符）
jsonschema.validate(instance=dplan, schema=schema)
print("結構層驗證通過！")
```

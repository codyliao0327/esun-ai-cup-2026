## 規則一 : 近5日相對市場報酬>0 以及 近10日相對市場報酬>0
#### 原始資料

| 中文名稱 | 程式變數 | 資料說明／單位 | 用途 |
|---|---|---|---|
| 股票清單 | `stocks` | 150檔的 `stock_id`、`stock`、`market` | 確定分析範圍與對應市場 |
| 個股收盤價 | `close` | `price:收盤價`，元 | 計算個股報酬率 |
| 市場價格指數 | `market_price_index` | 上市使用TAIEX，上櫃使用櫃買價格指數，均不含息 | 計算對應市場報酬率 |

#### 觀察指標

| 中文指標 | 程式變數／欄位 | 計算方式／單位 | 用途 |
|---|---|---|---|
| 個股近5日、10日報酬率 | `stock_returns` | （最新收盤價÷N個交易日前收盤價－1）×100，% | 觀察個股近期價格表現 |
| 市場近5日、10日報酬率 | `market_returns` | （最新市場指數÷N個交易日前指數－1）×100，% | 提供市場同期比較基準 |
| 近5日相對市場報酬 | `stock_summary["近5日相對市場報酬（百分點）"]` | 個股近5日報酬率－市場近5日報酬率，百分點 | 觀察近5日優於或落後市場的幅度 |
| 近10日相對市場報酬 | `stock_summary["近10日相對市場報酬（百分點）"]` | 個股近10日報酬率－市場近10日報酬率，百分點 | 觀察近10日優於或落後市場的幅度 |
| 規則一是否入選 | `stock_summary["規則一是否入選"]` | 兩項相對市場報酬皆大於0，`True`／`False` | 標示已確定納入股票池的股票 |

N分別為5與10，個股與市場使用共同交易日期。規則一要求兩項相對市場報酬同時大於0，不要求個股報酬一定為正。

```python
# 資料起始日期
START_DATE = "2026-10-01"

# 個股資料
close = data.get("price:收盤價", start=START_DATE)          # 一般收盤價

# 指數資料
index_close = data.get("stock_index_price:收盤指數", start=START_DATE)

# 大盤價格指數（不含息）
market_price_index = pd.concat({
    "twse": data.get(
        "taiex_total_index:收盤指數", start=START_DATE
    )["TAIEX"],
    "otc": index_close["上櫃櫃買指數:指數"],
}, axis=1)

# 使用共同日期，確保個股與市場比較相同期間
common_dates = close.index.intersection(market_price_index.index).sort_values()
close = close.reindex(index=common_dates)
market_price_index = market_price_index.reindex(index=common_dates)

assert len(common_dates) >= 11, "計算近10日報酬至少需要11個共同交易日"
print("價格資料日期：", common_dates[-1].date())
```

```python
"""計算截至最後一筆資料的近N個交易日報酬率（%）。"""
def calculate_return(prices, days):
    end_price = prices.iloc[-1]             # 最新日
    start_price = prices.iloc[-(days + 1)]  # N個交易日前
    return (end_price / start_price - 1) * 100
```

```python
stock_returns = pd.DataFrame({
    "近5日報酬率（%）": calculate_return(close, 5),
    "近10日報酬率（%）": calculate_return(close, 10),
})

market_returns = pd.DataFrame({
    "近5日市場報酬率（%）": calculate_return(market_price_index, 5),
    "近10日市場報酬率（%）": calculate_return(market_price_index, 10),
})
```

```python
# 3. 建立150檔規則一指標表
stock_summary = stocks.set_index("stock_id").join(stock_returns)
stock_summary = stock_summary.join(market_returns, on="market")

stock_summary.insert(
    stock_summary.columns.get_loc("market") + 1,
    "資料日期",
    common_dates[-1]
)

for days in [5, 10]:
    stock_summary[f"近{days}日相對市場報酬（百分點）"] = (
        stock_summary[f"近{days}日報酬率（%）"]
        - stock_summary[f"近{days}日市場報酬率（%）"]
    )

stock_summary["規則一是否入選"] = (
    (stock_summary["近5日相對市場報酬（百分點）"] > 0)
    & (stock_summary["近10日相對市場報酬（百分點）"] > 0)
)

display(stock_summary.round(4))
```



## 集保持股指標（週資料）

#### 原始資料

| 中文名稱 | 程式變數 | 單位 | 用途 |
|---|---|---|---|
| 大於一千張持股占比 | `holding_over_1000` | % | 計算千張占比變化 |
| 小於五十張持股占比 | `holding_under_50` | % | 計算小額占比變化 |
| 全部持有人數 | `holder_count` | 人 | 計算持有人數變化率 |

#### 觀察指標

| 中文指標 | 程式變數 | 計算方式／單位 | 用途 |
|---|---|---|---|
| 千張占比近1期變化 | `holding_over_1000_change` | `holding_over_1000.diff(1)`：本期占比減上期占比，百分點| 觀察大於一千張級距占比的增減 |
| 小於50張占比近1期變化 | `holding_under_50_change` | `holding_under_50.diff(1)`：本期占比減上期占比，百分點 | 觀察小於五十張級距占比的增減 |
| 持有人數近1期變化率 | `holder_change_1w` | （本期人數÷上期人數－1）×100，% | 觀察全部持有人數的相對增減幅度 |

近1期指相鄰兩筆集保週資料。

`chip_summary` 只輸出最新一期的三項變化指標，以及股票代碼、名稱、市場與資料日期。原始占比及人數用於計算，未另外輸出至此表。缺值不補0；人數變化率的上期人數為0時，結果保留缺值。

參考：
- 凱基：如何由籌碼的追蹤發現即將上漲的潛力股
- FinLab：散戶反向策略
- FinLab：籌碼訊號回測
- xq策略

```python
# 資料起訖日期
START_DATE = "2026-09-01"
END_DATE = pd.Timestamp.now(tz="Asia/Taipei").strftime("%Y-%m-%d")

# 股票代碼
stock_ids = stocks["stock_id"].tolist()


holding_over_1000 = data.get(
    "etl:inventory:大於一千張佔比",
    start=START_DATE, end=END_DATE
).reindex(columns=stock_ids)


holding_under_50 = data.get(
    "etl:inventory:小於五十張佔比",
    start=START_DATE, end=END_DATE
).reindex(columns=stock_ids)


holder_count = data.get(
    "etl:inventory:全部人數",
    start=START_DATE, end=END_DATE
).reindex(columns=stock_ids)
```

```python

holding_over_1000_change = holding_over_1000.diff(1)
holding_under_50_change = holding_under_50.diff(1)

holder_change_1w = (
    holder_count / holder_count.shift(1).replace(0, float("nan")) - 1
) * 100
```

```python
# 確認資料日期一致
assert holding_over_1000.index.equals(holding_under_50.index)
assert holding_over_1000.index.equals(holder_count.index)

# 建立最新一期總表
chip_summary = stocks[["stock_id", "stock", "market"]].copy()
chip_summary["資料日期"] = holding_over_1000.index[-1]

chip_summary["千張占比近1期變化（百分點）"] = (
    chip_summary["stock_id"].map(holding_over_1000_change.iloc[-1])
)

chip_summary["小於50張占比近1期變化（百分點）"] = (
    chip_summary["stock_id"].map(holding_under_50_change.iloc[-1])
)

chip_summary["持有人數近1期變化率（%）"] = (
    chip_summary["stock_id"].map(holder_change_1w.iloc[-1])
)

display(chip_summary.round(4))
```

## 融資指標（日資料）

#### 原始資料

| 中文名稱 | 程式變數 | 單位 | 用途 |
|---|---|---|---|
| 融資餘額 | `margin_balance` | 張 | 提供融資部位規模，計算變化率與近期百分位 |

#### 觀察指標

| 中文指標 | 程式變數／欄位 | 計算方式／單位 | 用途 |
|---|---|---|---|
| 最新融資餘額 | `margin_summary["融資餘額（張）"]` | 取最新一筆餘額，張 | 觀察目前融資部位規模 |
| 融資餘額近5日變化率 | `margin_change_5d` | （最新餘額÷5個交易日前餘額－1）×100，% | 觀察近期槓桿部位的增減幅度 |
| 融資餘額近10日百分位 | `margin_percentile_10d` | 最新餘額在含當日近10個交易日中的升冪排名÷10×100，同值採平均名次 | 觀察目前餘額在近期歷史的位置，越高表示位置越高 |

`margin_summary` 輸出最新一天的三項融資資料，以及股票代碼、名稱、市場與融資資料日期。缺值不補0；變化率基期為0時，結果保留缺值。

```python
# 資料起訖日期
START_DATE = "2026-09-01"
END_DATE = pd.Timestamp.now(tz="Asia/Taipei").strftime("%Y-%m-%d")

# 股票代碼
stock_ids = stocks["stock_id"].tolist()
margin_balance = data.get(
    "margin_transactions:融資今日餘額",
    start=START_DATE,
    end=END_DATE
).reindex(columns=stock_ids)
```

```python
#近5日變化率
margin_change_5d = (
    margin_balance
    / margin_balance.shift(5).replace(0, float("nan"))
    - 1
) * 100

# 融資餘額近10個交易日百分位
margin_percentile_10d = (
    margin_balance.rolling(10).rank(pct=True) * 100
)
```

```python
# 建立融資最新一天總表
margin_summary = stocks[["stock_id", "stock", "market"]].copy()

margin_summary["融資資料日期"] = margin_balance.index[-1]

margin_summary["融資餘額（張）"] = (
    margin_summary["stock_id"].map(margin_balance.iloc[-1])
)

margin_summary["融資餘額近5日變化率（%）"] = (
    margin_summary["stock_id"].map(margin_change_5d.iloc[-1])
)

margin_summary["融資餘額近10日百分位"] = (
    margin_summary["stock_id"].map(margin_percentile_10d.iloc[-1])
)

display(margin_summary.round(4))
```

## 本益比指標

#### 原始資料

| 中文名稱 | 程式變數 | 單位 | 用途 |
|---|---|---|---|
| 本益比 | `pe` | 倍 | 提供最新本益比，計算同產業百分位 |
| 產業分類 | `categories`、`industry` | 產業名稱 | 使用 `category` 欄位建立同產業比較群組 |

#### 觀察指標

| 中文指標 | 程式變數／欄位 | 計算方式／單位 | 用途 |
|---|---|---|---|
| 最新本益比 | `pe_summary["本益比"]` | 取最新一筆本益比，倍；不大於0時轉為缺值 | 觀察股價相對獲利的倍數 |
| 同產業本益比百分位 | `pe_summary["同產業本益比百分位"]` | （`pe_rank`－1）÷（`pe_count`－1）×100，同值採平均名次 | 觀察同產業的本益比位置，越低表示相對越低 |

`pe_rank` 為同產業本益比升冪排名，`pe_count` 為同產業有效家數，兩者只用於計算。

比較範圍為完整150檔內、同產業且本益比大於0的股票。只有一家有效資料時，百分位保留缺值。

`pe_summary` 輸出最新本益比、產業與同產業百分位，以及股票代碼、名稱、市場與本益比資料日期。


```python
START_DATE = "2026-09-01"
END_DATE = pd.Timestamp.now(tz="Asia/Taipei").strftime("%Y-%m-%d")

# 本益比
pe = data.get(
    "price_earning_ratio:本益比",
    start=START_DATE,
    end=END_DATE
).reindex(columns=stock_ids).sort_index()

# 產業分類
categories = data.get("security_categories")
```

```python
pe_summary = stocks[["stock_id", "stock", "market"]].copy()

pe_summary["本益比資料日期"] = pe.index[-1]

pe_summary["本益比"] = (
    pe_summary["stock_id"].map(pe.iloc[-1])
)

# 非正值不參與估值排名，保留缺值
pe_summary["本益比"] = pe_summary["本益比"].where(
    pe_summary["本益比"] > 0
)

# 加入產業
industry = categories[["stock_id", "category"]].drop_duplicates("stock_id")

pe_summary = pe_summary.merge(
    industry,
    on="stock_id",
    how="left"
).rename(columns={"category": "產業"})
```

```python
# 依產業計算本益比百分位
pe_group = pe_summary.groupby("產業")["本益比"]

pe_rank = pe_group.rank(method="average")
pe_count = pe_group.transform("count")

pe_summary["同產業本益比百分位"] = (
    (pe_rank - 1)
    / (pe_count - 1).replace(0, float("nan"))
    * 100
)

# 移除先前建立的排名與家數欄位
pe_summary = pe_summary.drop(
    columns=["同產業本益比排名", "同產業有效家數"],
    errors="ignore"
)

display(pe_summary.round(2))
```

## 資料日期與對照方式

各表依 `stock_id` 對照，保留各自的資料日期，不強制對齊日、週資料，也不使用前期資料補值。

| 資料表 | 程式變數 | 日期欄位 |
|---|---|---|
| 規則一指標表 | `stock_summary` | `資料日期` |
| 集保最新一期總表 | `chip_summary` | `資料日期` |
| 融資最新一天總表 | `margin_summary` | `融資資料日期` |
| 本益比最新一天總表 | `pe_summary` | `本益比資料日期` |
